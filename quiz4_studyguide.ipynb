### 3) FX conversions — % growth for EUR→USD and GBP→USD on one plot
Watch for

Use .diff() for the numerator and .shift() for “yesterday”.

Multiply by 100 to get percentage.

Plot both growth series together (either with pandas .plot() or plt.plot).

```python
# 1) read and sort
df = pd.read_csv("___fx.csv")
df["date"] = pd.to_datetime(df["actual name of date column"])
df = df.sort_values("date")

# 2) compute % growth (yesterday as denominator)
df["growth_eur"] = df["___eur_usd_col___"].diff() / df["___eur_usd_col___"].shift() * 100
df["growth_gbp"] = df["___gbp_usd_col___"].diff() / df["___gbp_usd_col___"].shift() * 100

# 3) plot together (either API is OK; this uses pandas .plot)
ax = df.plot(x="date", y=["growth_eur", "growth_gbp"], alpha=0.85)
ax.set_xlabel("Time")
ax.set_ylabel("Daily Percentage Change")
ax.set_title("FX Percentage Growth (EUR→USD vs GBP→USD)")
```

### 5) Highlight regions where the rate is above a value (plt.fill_between)
Watch for

Build a mask from the Boolean column.

Shade between the series and a baseline (0 is fine) where the mask is True.

Keep times sorted.

```python
#method 1
df = df.sort_values("date")
fig, ax = plt.subplots()                     # ← lecture style

ax.plot(df["date"], df["eur_usd"], label="EUR→USD")

thr = 1.20
plt.fill_between(                              # lecture used plt.fill_between
    df["date"], df["eur_usd"], 0,
    where=(df["eur_usd"] > thr),
    alpha=0.25
)

ax.set_xlabel("Time")
ax.set_ylabel("Rate")
ax.legend()
plt.show()

# method 2
# to use this method, df["eur_above"] must exist (created in Q4) and be bool.
# Date column converted with pd.to_datetime and sorted with sort_values("date").
df = df.sort_values("date")

fig, ax = plt.subplots()

ax.plot(df["date"], df["eur_usd"], label="EUR→USD")

plt.fill_between(
    df["date"],
    df["eur_usd"],
    0,                     # baseline
    where=df["eur_above"], # ← uses the Boolean column from Q4
    alpha=0.25
)

ax.set_xlabel("Time")
ax.set_ylabel("Rate")
ax.set_title("EUR→USD with high-rate regions highlighted")
ax.legend()
plt.show()
```
