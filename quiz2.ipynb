In general functions take the form:

```python

    #---- DEFINE
    def my_function(parameter):
        body
        return expression

    #---- RUN
    my_function(parameter = argument) 

    #---- RUN
    my_function(argument)
```
<br>

For multiple arguments or multiple return values:
```python

    #---- DEFINE
    def my_function(parameter1, parameter2, parameter3):
        body
        return expression1, expression2

    #---- RUN
    my_function(parameter1 = arg1, parameter2 = arg2, parameter3 = arg3) 

    #---- RUN
    my_function(arg1, arg2, arg3)
```

````python
#1
#Typical mistake (wrong return type used in string operation):
def label_sign(x):
    # ERROR: tried to concatenate a string and an int; str + int is invalid in Python
    return "positive: " + x

n = 3
print(label_sign(n))

# Fixed code
def label_sign(x):
    return "positive" if x > 0 else "non-positive"

n = 3
print(label_sign(n))

#(Other common gotchas: indenting return wrongly; using print instead of return; 
# forgetting to call the function; using a variable name that shadows the function name.)
````

````python
#2
#Typical mistake (list math does concatenation/repetition):
import numpy as np

xs = [1, 2, 3]
ys = [10, 20, 30]

# NOTE: with plain Python lists, xs + ys concatenates; xs * 2 repeats
# This runs, but gives the wrong result (concatenation), not elementwise addition.
wrong = xs + ys
print(wrong)

#Corrected (vectorize first):
import numpy as np

xs = [1, 2, 3]
ys = [10, 20, 30]

x_arr = np.array(xs)
y_arr = np.array(ys)

correct = x_arr + y_arr          # elementwise addition
# or: correct = x_arr * y_arr    # elementwise multiply
print(correct)

````

````python
#5
import numpy as np
import pandas as pd

# example DataFrame
df = pd.DataFrame({"id": range(1, 11)})

# 1) add random variable from a given distribution
# Uniform(0,1):
df["random_var"] = np.random.rand(len(df))
# OR Normal(mu, sigma):
# mu, sigma = 0, 1
# df["random_var"] = np.random.normal(mu, sigma, len(df))

# 2) split into two groups by a logical condition (example: <= 0.5 goes to A)
groupA = df.loc[df["random_var"] <= 0.5].copy()
groupB = df.loc[df["random_var"] > 0.5].copy()

print(groupA.head())
print(groupB.head())
````

````python
#6
import matplotlib.pyplot as plt

# Suppose both groups have columns x_col and y_col
# If not, rename to whatever your prompt says.
x_col = "xval"
y_col = "yval"

# Example toy data if needed:
# groupA[x_col] = ...
# groupA[y_col] = ...
# groupB[x_col] = ...
# groupB[y_col] = ...

plt.figure()
plt.scatter(groupA[x_col], groupA[y_col], label="Group A", marker="o")
plt.scatter(groupB[x_col], groupB[y_col], label="Group B", marker="x")

plt.xlabel(x_col)
plt.ylabel(y_col)
plt.title("Scatter by Random Assignment")
plt.legend()
plt.show()
````

````python
# Q5: random assignment into two groups
np.random.seed(42)                         # reproducible randomness (optional)
df = df.copy()                             # avoid modifying original, if needed
df["random_var"] = np.random.rand(len(df)) # Uniform(0,1)

# Split into two groups by a logical condition on random_var
groupA = df[df["random_var"] < 0.5]        # Group A
groupB = df[df["random_var"] >= 0.5]       # Group B

# Q6: side-by-side scatter plots using subplots
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10,4), sharex=True, sharey=True)

# Left plot (Group A)
ax[0].scatter(groupA["x_col"], groupA["y_col"], s=18, alpha=0.7)


# Right plot (Group B)
ax[1].scatter(groupB["x_col"], groupB["y_col"], s=18, alpha=0.7)
ax[1].set_title("Group B (rand ≥ 0.5)")
ax[1].set_xlabel("x_col")

fig.suptitle("Random Assignment: Side-by-Side Scatter")
fig.tight_layout()
plt.show() 
````
