### Question 2: Build a NumPy array from two lists + histogram (with custom color). ### Create two lists x_vals and y_vals of the same length. Make a NumPy array z = 2*np.array(x_vals) - np.array(y_vals). Plot a histogram of z with: bins=10, labeled axes + title, a non-default bar color (e.g., "orange"). Show the plot

```python
x_vals = [2, 4, 6, 8, 10, 12, 14]
y_vals = [1, 1, 3, 3, 5, 5, 7]

z = 2*np.array(x_vals) - np.array(y_vals)

plt.figure(figsize=(6,4))
plt.hist(z, bins=10, edgecolor="black", color="orange")
plt.title("Histogram of z = 2x - y")
plt.xlabel("z value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

### Question 3. Random samples from two dists + side-by-side histograms
### Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. Each sub-figure will have its own labeled axes and title.
### Set a seed for reproducibility. Draw 1,000 samples from Normal(0,1) and 1,000 from Uniform(0,1). Make one figure with two subplots (1 row × 2 columns). Each subplot: histogram with bins=30, its own title, labeled axes. Use sharey=True so heights are comparable. Show the figure. 

```python
np.random.seed(1234)

# generate samples
normal_samples = np.random.normal(0, 1, 1000)
expo_samples = np.random.exponential(1, 1000)

# first histogram: normal
plt.hist(normal_samples, bins=30)
plt.title("Normal(0,1)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# second histogram: exponential
plt.hist(expo_samples, bins=30)
plt.title("Exponential(1)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
